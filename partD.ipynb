## Data Preprocessing and Model Training

```python
# Create training dataset
df = create_training_data()
print(f"Training dataset shape: {df.shape}")
print("\nAlgorithm distribution:")
print(df['algorithm'].value_counts())
print("\nSample data:")
df.head()
```

```python
# Encode categorical features
label_encoders = {}
X_encoded = pd.DataFrame()

for column in df.columns[:-1]:  # Exclude target 'algorithm' column
    le = LabelEncoder()
    X_encoded[column] = le.fit_transform(df[column])
    label_encoders[column] = le

# Encode target variable
target_encoder = LabelEncoder()
y_encoded = target_encoder.fit_transform(df['algorithm'])

print(f"Features shape: {X_encoded.shape}")
print(f"Target shape: {y_encoded.shape}")
print(f"\nAlgorithm classes: {list(target_encoder.classes_)}")
```

```python
# Split data for training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

# Train decision tree
dt_classifier = DecisionTreeClassifier(
    max_depth=5, 
    min_samples_split=2, 
    min_samples_leaf=1, 
    random_state=42
)

dt_classifier.fit(X_train, y_train)

# Evaluate model
y_pred = dt_classifier.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"Decision Tree Accuracy: {accuracy:.3f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=target_encoder.classes_))
```

## Algorithm Selection Function

```python
def recommend_algorithm(dataset_size, n_features, linearity, problem_type, 
                        n_labels, feature_types, interpretability):
    """
    Recommend the best algorithm based on dataset characteristics.
    
    Parameters:
    -----------
    dataset_size : str
        'small' (<1K samples), 'medium' (1K-100K), 'large' (>100K)
    n_features : str
        'low' (<10), 'medium' (10-100), 'high' (>100)
    linearity : str
        'linear' or 'non_linear'
    problem_type : str
        'classification', 'clustering', or 'dimensionality_reduction'
    n_labels : str
        'binary', 'multiclass', or 'none' (for clustering/reduction)
    feature_types : str
        'numerical', 'categorical', or 'mixed'
    interpretability : str
        'high', 'medium', or 'low'
    
    Returns:
    --------
    dict
        Recommended algorithm and confidence score
    """
    
    # Prepare input data
    input_data = pd.DataFrame({
        'dataset_size': [dataset_size],
        'n_features': [n_features],
        'linearity': [linearity],
        'problem_type': [problem_type],
        'n_labels': [n_labels],
        'feature_types': [feature_types],
        'interpretability': [interpretability]
    })
    
    # Encode input using trained label encoders
    input_encoded = pd.DataFrame()
    for column in input_data.columns:
        if column in label_encoders:
            # Handle unknown categories
            if input_data[column].iloc[0] not in label_encoders[column].classes_:
                # Use most common value as fallback
                input_encoded[column] = [0]
            else:
                input_encoded[column] = label_encoders[column].transform(input_data[column])
        else:
            input_encoded[column] = [0]
    
    # Make prediction
    prediction = dt_classifier.predict(input_encoded)[0]
    probabilities = dt_classifier.predict_proba(input_encoded)[0]
    
    # Get algorithm name and confidence
    recommended_algorithm = target_encoder.inverse_transform([prediction])[0]
    confidence = probabilities[prediction]
    
    # Get top 3 alternatives
    top_3_idx = np.argsort(probabilities)[-3:][::-1]
    alternatives = []
    for idx in top_3_idx:
        if idx != prediction:
            alternatives.append({
                'algorithm': target_encoder.inverse_transform([idx])[0],
                'confidence': probabilities[idx]
            })
    
    return {
        'recommended_algorithm': recommended_algorithm,
        'confidence': confidence,
        'alternatives': alternatives[:2],  # Top 2 alternatives
        'input_characteristics': {
            'dataset_size': dataset_size,
            'n_features': n_features,
            'linearity': linearity,
            'problem_type': problem_type,
            'n_labels': n_labels,
            'feature_types': feature_types,
            'interpretability': interpretability
        }
    }
```

## Interactive Algorithm Selector

```python
def interactive_algorithm_selector():
    """
    Interactive function to guide users through algorithm selection
    """
    print("=== Algorithm Selection Assistant ===")
    print("Answer the following questions to get the best algorithm recommendation:\n")
    
    # Collect user input
    questions = {
        'dataset_size': {
            'question': 'What is your dataset size?',
            'options': ['small (<1K samples)', 'medium (1K-100K)', 'large (>100K)'],
            'values': ['small', 'medium', 'large']
        },
        'n_features': {
            'question': 'How many features does your dataset have?',
            'options': ['low (<10)', 'medium (10-100)', 'high (>100)'],
            'values': ['low', 'medium', 'high']
        },
        'linearity': {
            'question': 'What is the relationship between features and target?',
            'options': ['linear', 'non-linear'],
            'values': ['linear', 'non_linear']
        },
        'problem_type': {
            'question': 'What type of problem are you solving?',
            'options': ['classification', 'clustering', 'dimensionality_reduction'],
            'values': ['classification', 'clustering', 'dimensionality_reduction']
        },
        'n_labels': {
            'question': 'How many target labels/classes?',
            'options': ['binary (2 classes)', 'multiclass (>2 classes)', 'none (clustering/reduction)'],
            'values': ['binary', 'multiclass', 'none']
        },
        'feature_types': {
            'question': 'What types of features do you have?',
            'options': ['numerical only', 'categorical only', 'mixed numerical/categorical'],
            'values': ['numerical', 'categorical', 'mixed']
        },
        'interpretability': {
            'question': 'How important is model interpretability?',
            'options': ['high (very important)', 'medium (somewhat important)', 'low (not important)'],
            'values': ['high', 'medium', 'low']
        }
    }
    
    user_inputs = {}
    
    for key, q_info in questions.items():
        print(f"\n{q_info['question']}")
        for i, option in enumerate(q_info['options'], 1):
            print(f"{i}. {option}")
        
        while True:
            try:
                choice = int(input("Enter your choice (number): "))
                if 1 <= choice <= len(q_info['options']):
                    user_inputs[key] = q_info['values'][choice - 1]
                    break
                else:
                    print("Invalid choice. Please try again.")
            except ValueError:
                print("Please enter a valid number.")
    
    # Get recommendation
    result = recommend_algorithm(**user_inputs)
    
    # Display results
    print("\n" + "="*50)
    print("ALGORITHM RECOMMENDATION")
    print("="*50)
    print(f"\n🎯 Recommended Algorithm: {result['recommended_algorithm']}")
    print(f"📊 Confidence: {result['confidence']:.1%}")
    
    if result['alternatives']:
        print("\n🔄 Alternative Options:")
        for i, alt in enumerate(result['alternatives'], 1):
            print(f"   {i}. {alt['algorithm']} ({alt['confidence']:.1%} confidence)")
    
    print("\n📋 Your Dataset Characteristics:")
    for key, value in result['input_characteristics'].items():
        print(f"   • {key.replace('_', ' ').title()}: {value}")
    
    return result

# Uncomment to run interactive selector
# result = interactive_algorithm_selector()
```
